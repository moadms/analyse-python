# 🔍 ANALYSE DU DÉSÉQUILIBRE DES CLASSES (Stroke Dataset)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Charger les données
df = pd.read_csv('healthcare-dataset-stroke-data.csv')

# Analyser le déséquilibre
print("="*60)
print("📊 ANALYSE DU DÉSÉQUILIBRE DES CLASSES")
print("="*60)
print("\n1️⃣ Distribution de la variable cible (stroke):")
print(df['stroke'].value_counts())
print("\n2️⃣ Pourcentages:")
print(df['stroke'].value_counts(normalize=True) * 100)

# Calculer le ratio de déséquilibre
class_0 = df['stroke'].value_counts()[0]
class_1 = df['stroke'].value_counts()[1]
ratio = class_0 / class_1

print(f"\n3️⃣ Ratio de déséquilibre: {ratio:.2f}:1")
print(f"   👉 Il y a {ratio:.0f}x plus de personnes SAINES que MALADES")
print(f"   👉 Classe 0 (Sain): {class_0} personnes")
print(f"   👉 Classe 1 (Stroke): {class_1} personnes")

# Visualisation
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Graphique 1: Count plot
sns.countplot(data=df, x='stroke', ax=axes[0], palette=['#2ecc71', '#e74c3c'])
axes[0].set_title('Distribution des Classes (Stroke)', fontsize=14, fontweight='bold')
axes[0].set_xlabel('Stroke (0=Sain, 1=Malade)', fontsize=12)
axes[0].set_ylabel('Nombre de patients', fontsize=12)
for container in axes[0].containers:
    axes[0].bar_label(container)

# Graphique 2: Pie chart
colors = ['#2ecc71', '#e74c3c']
explode = (0, 0.1)
axes[1].pie(df['stroke'].value_counts(), 
            labels=['Sain (0)', 'Stroke (1)'], 
            autopct='%1.1f%%',
            colors=colors,
            explode=explode,
            shadow=True,
            startangle=90)
axes[1].set_title('Proportion des Classes', fontsize=14, fontweight='bold')

plt.tight_layout()
plt.show()
```

**Variable : bmi (Quantitative)**

Valeurs manquantes

```python
df['bmi'].isnull().sum()
```

Remplacement par la médiane

```python
df['bmi'] = df['bmi'].fillna(df['bmi'].median())

```

Boxplot + IQR

```python
sns.boxplot(x=df['bmi'])
plt.show()
```

```python
Q1 = df['bmi'].quantile(0.25)
Q3 = df['bmi'].quantile(0.75)
IQR = Q3 - Q1

df = df[(df['bmi'] >= Q1 - 1.5*IQR) & (df['bmi'] <= Q3 + 1.5*IQR)]
```

QQPLOT

```python
stats.probplot(df['bmi'], plot=plt)
plt.title("QQPLOT BMI")
plt.show()
```

La variable bmi contient des valeurs manquantes, remplacées par la médiane, méthode robuste face aux valeurs extrêmes.
Les valeurs aberrantes ont été traitées par la méthode IQR en supprimant les individus extrêmes, afin d’améliorer la stabilité statistique de la variable.
Après traitement, la distribution reste légèrement asymétrique.

#Cellule 4 - Analyse bi-variable

**Sélection des variables quantitatives explicatives**

```python
variables_quantitatives = ['age', 'avg_glucose_level', 'bmi']
df_quant = df[variables_quantitatives]
df_quant.head()
```

**Matrice de corrélation de Pearson**

```python
corr_matrix = df_quant.corr(method='pearson')
corr_matrix
```

**Heatmap de corrélation**

```python
plt.figure(figsize=(6,4))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Matrice de corrélation de Pearson")
plt.show()
```

La matrice de corrélation de Pearson montre que les coefficients de corrélation entre les variables quantitatives sont inférieurs à 0.6.
Cela indique l’absence de corrélations linéaires fortes entre les variables explicatives.
Les relations observées sont faibles à modérées, ce qui signifie que les variables apportent des informations complémentaires et ne présentent pas de multicolinéarité importante.

# 🔷 II-4-1 — Cas des variables numériques (ACP)
✅ 1. Vérifier les conditions d’application de l’ACP
📌 Conditions :

Variables numériques ✅

Variables corrélées (au moins un peu)

Données centrées-réduites (obligatoire)

🔹 Variables numériques

```python
variables_quantitatives = ['age', 'avg_glucose_level', 'bmi']
df_num = df[variables_quantitatives]
```

🔹 Vérifier les corrélations

```python
df_num.corr()
```

La matrice de corrélation montre l’existence de corrélations positives faibles à modérées entre les variables quantitatives.
La corrélation la plus élevée est observée entre l’âge et le BMI (r = 0,38), indiquant que les individus plus âgés ont tendance à présenter un BMI plus élevé.
Les corrélations entre le niveau moyen de glucose et les autres variables sont faibles, ce qui signifie que cette variable apporte une information complémentaire.
Ces résultats justifient l’application de l’Analyse en Composantes Principales (ACP).

# ✅ 2. Centrage – Réduction des données

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_num)
```

# ✅ 3. Appliquer l’ACP

```python
from sklearn.decomposition import PCA

pca = PCA()
X_pca = pca.fit_transform(X_scaled)
```

# ✅ 4. Pourcentage de variance expliquée

```python
pca.explained_variance_ratio_
```

```python
pca.explained_variance_ratio_.cumsum()
```

📊 Graphique : Éboulis des valeurs propres

```python
plt.plot(range(1, len(pca.explained_variance_)+1),
         pca.explained_variance_ratio_,
         marker='o')
plt.xlabel("Composantes principales")
plt.ylabel("Variance expliquée")
plt.title("Scree Plot")
plt.show()
```

# ✅ 5. Cercle de corrélation (interprétation des variables)

```python
components = pca.components_

plt.figure(figsize=(6,6))
for i, var in enumerate(variables_quantitatives):
    plt.arrow(0, 0, components[0,i], components[1,i],
              head_width=0.05)
    plt.text(components[0,i]*1.1,
             components[1,i]*1.1,
             var)

plt.axhline(0)
plt.axvline(0)
plt.xlabel("Dim 1")
plt.ylabel("Dim 2")
plt.title("Cercle de corrélation")
plt.grid()
plt.show()
```

# ✅ 6. Projection des individus

```python
plt.scatter(X_pca[:,0], X_pca[:,1], alpha=0.5)
plt.xlabel("Dim 1")
plt.ylabel("Dim 2")
plt.title("Projection des individus")
plt.show()
```

# 🔷 II-4-2 — Cas des variables catégorielles (ACM)
🎯 Objectif de l’ACM

👉 Étudier les relations entre modalités
👉 Comprendre les profils d’individus

🔹 Variables catégorielles

```python
variables_cat = [
    'gender',
    'ever_married',
    'work_type',
    'Residence_type',
    'smoking_status'
]

df_cat = df[variables_cat]
```

# ✅ 1. Encodage disjonctif (one-hot)

```python
df_dummies = pd.get_dummies(df_cat)
```

# ✅ 2. Appliquer l’ACM

```python
from sklearn.decomposition import PCA

acm = PCA(n_components=2)
X_acm = acm.fit_transform(df_dummies)
```

# ✅ 3. Variance expliquée

```python
acm.explained_variance_ratio_
```

📊 Projection des individus (ACM)

```python
plt.scatter(X_acm[:,0], X_acm[:,1], alpha=0.5)
plt.xlabel("Dim 1")
plt.ylabel("Dim 2")
plt.title("ACM - Projection des individus")
plt.show()
```

📊 Projection des modalités

```python
loadings = acm.components_.T

plt.figure(figsize=(8,6))
for i, col in enumerate(df_dummies.columns):
    plt.scatter(loadings[i,0], loadings[i,1])
    plt.text(loadings[i,0], loadings[i,1], col, fontsize=8)

plt.axhline(0)
plt.axvline(0)
plt.title("ACM - Modalités")
plt.grid()
plt.show()
```

Les individus fumeurs et travaillant dans le secteur privé ont tendance à être proches dans l’espace factoriel.
Les individus mariés et résidant en zone urbaine présentent des profils similaires.
